```python
#Sofia
url = '/content/drive/MyDrive/Fundamentos de machine learning/Anexo ET_demo_round_traces_2022.csv'
df = pd.read_csv(url, sep=";", encoding='utf-8')
```

```python
#Con esta linea de codigo podemos asegurarnos de que las variables a utilizar sean numericas
df['MatchAssists'] = pd.to_numeric(df['MatchAssists'], errors='coerce')
df['TimeAlive'] = pd.to_numeric(df['TimeAlive'], errors='coerce')
df['MatchHeadshots'] = pd.to_numeric(df['MatchHeadshots'], errors='coerce')
df['RoundKills'] = pd.to_numeric(df['RoundKills'], errors='coerce')
df['PrimarySniperRifle'] = df['PrimarySniperRifle'].round().astype(int)
df['PrimaryAssaultRifle'] = df['PrimaryAssaultRifle'].round().astype(int)
df['PrimaryHeavy'] = df['PrimaryHeavy'].round().astype(int)
df['PrimarySMG'] = df['PrimarySMG'].round().astype(int)
```

```python
# 3.1. Imputar NaN en columnas numéricas usando la media
num_cols = df.select_dtypes(include=[np.number]).columns

for col in num_cols:
    if df[col].isnull().sum() > 0:
        mean_value = df[col].mean()
        df[col].fillna(mean_value, inplace=True)
```

```python
# 3.2.En columnas categóricas rellenamos con el valor más frecuente
cat_cols = df.select_dtypes(include=['object']).columns

for col in cat_cols:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()[0]
        df[col].fillna(mode_value, inplace=True)
```

```python
print("\nValores nulos por columna luego de añadir el valor segun la media:")
print(df.isnull().sum())
```

```python
# --- IMPORTAR LIBRERÍAS ---
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix

# --- 1. LIMPIEZA Y FILTRADO INICIAL ---
df['Survived'] = df['Survived'].astype(str).str.strip()
df = df[df['Survived'].isin(['True', 'False'])].copy()
df['Survived'] = df['Survived'].map({'True': 1, 'False': 0})
df = df[df['AbnormalMatch'] == False]

# --- 2. SELECCIÓN DE VARIABLES PARA MODELO ---
X = df[['RLethalGrenadesThrown', 'RNonLethalGrenadesThrown', 'RoundKills']].copy()
y = df['Survived']


# --- 4. DIVISIÓN EN ENTRENAMIENTO Y TEST ---
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
```

```python
# ------------------ ENTRENAMIENTO SVM ------------------
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
import seaborn as sns
import matplotlib.pyplot as plt

modelo_svm = SVC(kernel='linear', probability=True, random_state=42)
modelo_svm.fit(X_train, y_train)

# ------------------ PREDICCIÓN ------------------
y_pred_svm = modelo_svm.predict(X_test)
print(classification_report(y_test, y_pred_svm))

# Matriz de confusión
matriz_svm = confusion_matrix(y_test, y_pred_svm)
sns.heatmap(matriz_svm, annot=True, fmt='d', cmap='Blues')
plt.title("Matriz de Confusión - SVM")
plt.xlabel("Predicción")
plt.ylabel("Valor Real")
plt.show()
```

```python
# Curva ROC
y_prob_svm = modelo_svm.predict_proba(X_test)[:, 1]
fpr, tpr, _ = roc_curve(y_test, y_prob_svm)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, label=f"ROC Curve (AUC = {roc_auc:.2f})", color="orange")
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel("FPR")
plt.ylabel("TPR")
plt.title("Curva ROC - SVM")
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```
